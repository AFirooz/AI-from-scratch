```python
def lomuto_partition(array: list, start_idx=None, end_idx=None) -> tuple[int, list]:
    """
    Partitions a subarray `array[start_idx ... end_idx]` using the Lomuto scheme.
    Selects the first element `array[start_idx]` as the pivot.
    lomuto_idx is the index where:
        1. All values on the right of lomuto_idx are bigger than lomuto_value, and
        2. All values on the left of lomuto_idx are smaller than lomuto_value.
    """
    # init
    start_idx = 0 if not start_idx else start_idx
    end_idx = len(array) if not end_idx else end_idx

    lomuto_value = array[start_idx]  # holds the value, but does not reference the array's memory location
    lomuto_idx = start_idx           # tracks the boundary of elements strictly less than lomuto_value

    for i in range(start_idx+1, end_idx):
        if lomuto_value > array[i]:
            lomuto_idx += 1
            # Swap the values
            # Pythonic in-place swapping
            array[lomuto_idx], array[i] = array[i], array[lomuto_idx] 

    # Move the actual pivot element from the start to its final boundary position
    array[start_idx], array[lomuto_idx] = array[lomuto_idx], array[start_idx]
    
    return lomuto_idx, array
```

```python
idx, new_array = lomuto_partition([6, 4, 13, 11, 3, 12])

print(f'{idx=}')
new_array
```
