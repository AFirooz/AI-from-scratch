```python
def lomuto_partition(array: list, start_idx=None, end_idx=None, debug=True) -> tuple[int, list]:
    """
    Partitions a subarray `array[start_idx ... end_idx]` using the Lomuto scheme.
    Selects the first element `array[start_idx]` as the pivot.
    """
    if debug: print("START LOMUTO")
    
    start_idx = 0 if start_idx is None else start_idx
    end_idx = len(array) if end_idx is None else end_idx

    pivot_value = array[start_idx]
    pivot_idx = start_idx

    if debug: print(f"Pivot: array[{start_idx}] = {pivot_value}\n")

    for i in range(start_idx + 1, end_idx):
        if debug: print(f"Checking array[{i}]={array[i]} ", end="")
        if array[i] < pivot_value:
            pivot_idx += 1
            if debug:
                print(f"< pivot:array[{pivot_idx}]={pivot_value} --> "
                      f"Expanding the \"smaller than pivot\" zone by incrementing the pivot_idx to {pivot_idx}"
                      f"\nSwapping array[{pivot_idx}]={array[pivot_idx]} with array[{i}]={array[i]}")
            # Pythonic in-place swapping
            array[pivot_idx], array[i] = array[i], array[pivot_idx]
            if debug: 
                print(f"New array={array}")
                if array[pivot_idx] == array[i]: print(f'*  array[{i}]={array[i]} is already in the correct location')
                print()
        else:
            if debug: print(f">= pivot:array[{pivot_idx}]={pivot_value} --> Skipping")

    # Final pivot placement
    if debug: print(f"Finally, moving pivot {pivot_value} to final boundary {pivot_idx} of the \"smaller than pivot\" zone")
    array[start_idx], array[pivot_idx] = array[pivot_idx], array[start_idx]
    
    if debug: print("END LOMUTO")
    return pivot_idx, array
```

```python
# array = [6, 4, 13, 11, 3, 12]
array = [4, 1, 10, 8, 7, 12, 9, 2, 15]

idx, new_array = lomuto_partition(array)
print(f'{idx=}')
new_array
```

```python
def quick_select(array:list, k_smallest:int, start_idx=None, end_idx=None) -> tuple[int, list]:
    """Finds the k-th smallest element in `array` (k is 0-indexed global target)."""

    # init
    start_idx = 0 if start_idx is None else start_idx
    end_idx = len(array) if end_idx is None else end_idx

    print(f"array={array[start_idx:end_idx]}")

    # Base case: Search space is reduced to a single element
    if end_idx - start_idx <= 1:
        return array[start_idx]

    # Unpack the tuple returned by your custom lomuto_partition
    lomuto_idx, lomuto_array = lomuto_partition(array, start_idx, end_idx, debug=False)

    print(f"{k_smallest=} \n{lomuto_idx=} \nlomuto_array={lomuto_array[start_idx:end_idx]}")

    if k_smallest == lomuto_idx:
        print("k_smallest == lomuto_idx  -->  found it")
        return array[lomuto_idx], array
    
    elif k_smallest < lomuto_idx:
        # Search left: lomuto_idx becomes the new exclusive end_idx
        print("k_smallest < lomuto_idx  -->  going left \n")
        return quick_select(array, k_smallest, start_idx, lomuto_idx)
    
    else:  # k_smallest > lomuto_idx
        # Search right: start immediately after lomuto_idx
        # because we already know lomuto_idx is not it
        print("k_smallest > lomuto_idx  -->  going right \n")
        return quick_select(array, k_smallest, lomuto_idx + 1, end_idx)
```
