```python
def hoare_partition(arr, start=None, end=None, debug=True):
    """
    Partitions a subarray `arr[start ... end-1]` using Hoare's scheme.
    Unlike Lomuto, the pivot is not guaranteed to be at the returned index `j` in the sorted array.
    The partitions are guaranteed to be `arr[start ... j] <= pivot` and `arr[j+1 ... end-1] >= pivot`.
    Note that the pivot can be in both zones.
    """
    if debug:
        print("START HOARE")

    start = 0 if start is None else start
    end = len(arr) if end is None else end

    pivot = arr[start]
    i = start - 1
    j = end

    if debug:
        print(f"Pivot: array[{start}]={pivot}\n")

    while True:
        # Scan right looking for an element >= pivot
        while True:
            i += 1
            if debug:
                print(f"i-scan: checking arr[{i}]={arr[i]}", end="")
            if arr[i] >= pivot:
                if debug: print(f" --> i stopped at array[{i}]={arr[i]} >= pivot({pivot})")
                break
            else:
                if debug: print(" --> Skipping")

        # Scan left looking for an element <= pivot
        while True:
            j -= 1
            if debug:
                print(f"j-scan: checking arr[{j}] = {arr[j]}", end="")
            if arr[j] <= pivot:
                if debug:print(f" --> j stopped at arr[{j}] <= pivot({pivot})")
                break
            else:
                if debug: print(" --> Skipping")

        # If pointers cross or meet, the partitioning is complete
        if i >= j:
            if debug:
                print(f"\nPointers crossed (i={i}, j={j})")
                print("END HOARE")
            return j, arr

        # Swap elements that are on the wrong side of the pivot
        if debug:
            print(f"\nSwapping array[{i}]={arr[i]} with array[{j}]={arr[j]}")
        arr[i], arr[j] = arr[j], arr[i]
        if debug:
            print(f"New array: {arr}\n")
```

```python
array = [4, 1, 10, 8, 7, 12, 9, 2, 15]

j, arr = hoare_partition(array)

print(f"{j=}")
arr
```
